```python
# Dependency/path fixing for ColabFold
import os, sys, types

tf_lite_dir = "/usr/local/lib/python3.12/dist-packages/tensorflow/lite"

# 1. Find and delete ALL broken .so files under tensorflow/lite
deleted = []
for root, dirs, files in os.walk(tf_lite_dir):
    for fname in files:
        if fname.endswith(".so"):
            fpath = os.path.join(root, fname)
            os.remove(fpath)
            deleted.append(fpath)

print(f"Deleted {len(deleted)} .so files:")
for p in deleted:
    print(" ", p)

# 2. Stub the entire tensorflow.lite namespace in sys.modules
#    so any subprocess that imports tensorflow gets a no-op lite module
stub_init = """\
# Auto-stubbed: all .so files were broken (undefined symbol errors)
class _Stub:
    def __init__(self, *a, **kw): pass
    def __call__(self, *a, **kw): return self
    def __getattr__(self, name): return self

class MetricsWrapper(_Stub): pass
class Interpreter(_Stub): pass

def __getattr__(name):
    return _Stub()
"""

# Walk the lite dir and write __init__.py stubs for every subpackage
for root, dirs, files in os.walk(tf_lite_dir):
    init_path = os.path.join(root, "__init__.py")
    # Only overwrite if it's a pure-stub situation (no valid .py logic needed)
    # We append a safe __getattr__ fallback to existing inits
    if os.path.exists(init_path):
        with open(init_path, "a") as f:
            f.write("\n\ndef __getattr__(name):\n    class _S:\n        def __getattr__(self,n): return _S()\n        def __call__(self,*a,**k): return _S()\n    return _S()\n")
    else:
        with open(init_path, "w") as f:
            f.write(stub_init)

# 3. Also stub the specific modules that were individually broken
broken_modules = [
    ("tensorflow.lite.python.metrics._pywrap_tensorflow_lite_metrics_wrapper", "MetricsWrapper"),
    ("tensorflow.lite.python.interpreter_wrapper._pywrap_tensorflow_interpreter_wrapper", None),
]
for mod_path, attr in broken_modules:
    stub_dir = "/usr/local/lib/python3.12/dist-packages/" + mod_path.replace(".", "/")
    parent_dir = os.path.dirname(stub_dir)
    stub_file = os.path.join(
        "/usr/local/lib/python3.12/dist-packages",
        *mod_path.split(".")[:-1]
    )
    fname = mod_path.split(".")[-1] + ".py"
    fpath = os.path.join(
        "/usr/local/lib/python3.12/dist-packages",
        *mod_path.split(".")[:-1],
        fname
    )
    with open(fpath, "w") as f:
        if attr:
            f.write(f"class {attr}:\n    pass\n\ndef __getattr__(name):\n    return type(name, (), )()\n")
            f.write("def __getattr__(name):\n    return type(name, (), {})()\n")
    print(f"Wrote stub: {fpath}")

print("\nDone. Run colabfold_batch now.")
```

```python
import os
from pathlib import Path

# --- CONFIGURATION ---
OLD_PEPTIDE = "HIYTHMSHFIKQCFSLP"
OLD_A3M_PATH = "00_cd109_HIYTHMSHFIKQCFSLP.a3m"
# Where we will put the 49 new offline files
NEW_A3M_DIR = Path("offline_msas")
NEW_A3M_DIR.mkdir(exist_ok=True)
FASTA_DIR = Path("input")

# 1. Read the lead MSA
with open(OLD_A3M_PATH, 'r') as f:
    golden_a3m_content = f.read()

# 2. Extract  new candidates from the FASTAs and generate new MSAs
fasta_files = list(FASTA_DIR.glob("*.fa*"))
success_count = 0

for fa in fasta_files:
    with open(fa, 'r') as f:
        content = f.read().split('\n')
        name = content[0].replace('>', '').strip()
        # Extract just the 17-mer
        full_seq = "".join(content[1:]).strip()
        new_peptide = full_seq.split(':')[-1] if ':' in full_seq else full_seq[-17:]

    # Hijack the MSA by swapping the peptide sequence exactly
    new_a3m_content = golden_a3m_content.replace(OLD_PEPTIDE, new_peptide)

    # Save the new a3m file ready for offline folding
    out_path = NEW_A3M_DIR / f"{name}.a3m"
    with open(out_path, 'w') as f:
        f.write(new_a3m_content)
    success_count += 1

print(f"[+] Successfully  generated {success_count}  MSAs.")
```
